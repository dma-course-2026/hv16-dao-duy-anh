```python
#bai5
def in_chuoi_dai_hon(s1, s2):
    if len(s1) > len(s2):
        print(s1)
    elif len(s2) > len(s1):
        print(s2)
    else:
        print(s1)
        print(s2)
in_chuoi_dai_hon(input("Nhập chuỗi 1: "), input("Nhập chuỗi 2: "))
```

```python
#bai27
import math
class Circle:
    def __init__(self, r):
        self.r = r
    def dien_tich(self):
        return math.pi * self.r ** 2
c = Circle(5)
print(c.dien_tich())
```

```python
#bai28
class Hinhchunhat:
    def __init__(self, dai, rong):
        self.dai = dai
        self.rong = rong
    def dien_tich(self):
        return self.dai * self.rong
h = Hinhchunhat(6, 4)
print(h.dien_tich())
```

```python
#bai29
class Shape:
    def __init__(self):
        pass
    def area(self):
        print(0)
class Square(Shape):
    def __init__(self, length=0):
        Shape.__init__(self)
        self.length = length
    def area(self):
        print(self.length * self.length)
Shape().area()
Square(4).area()
```
