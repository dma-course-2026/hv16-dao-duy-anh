```python
#bai9
from datetime import datetime

ns = input("Nhập ngày sinh (y/m/d): ")
sinh = datetime.strptime(ns, "%Y/%m/%d")
hn = datetime.now()
tuoi = hn.year - sinh.year - ((hn.month, hn.day) < (sinh.month, sinh.day))
print("Tuổi:", tuoi)
```

```python
#bai10
gio = float(input("Số giờ làm mỗi tuần: "))
luong = float(input("Thù lao mỗi giờ chuẩn: "))
if gio <= 44:
    tien = gio * luong
else:
    tien = 44 * luong + (gio - 44) * luong * 1.5
print("Thực lĩnh:", tien)
```

```python
#bai23
def kiem_tra_doi_xung(a):
    return a == a[::-1]


def sx_doi_xung(a):
    b = a[:]
    n = len(b)
    for i in range(n // 2):
        if b[i] != b[n - 1 - i]:
            b[n - 1 - i] = b[i]
    return b


a = list(map(int, input("Nhập dãy số: ").split()))
if kiem_tra_doi_xung(a):
    print("Dãy đối xứng:", a)
else:
    print("Dãy chưa đối xứng:", a)
    print("Dãy sau khi biến đổi:", sx_doi_xung(a))
```
